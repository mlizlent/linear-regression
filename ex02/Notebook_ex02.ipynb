```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression

X, y, coef = make_regression(
    n_samples=100,
    n_features=1,
    n_informative=1,
    noise=10,
    coef=True,
    random_state=0,
    bias=100.0
)
plt.figure(figsize=(8, 5))
plt.scatter(X, y)
plt.xlabel("X")
plt.ylabel("y")
plt.title("X(1 dimension) and y")
plt.show()
```

```python
model = LinearRegression()
model.fit(X, y)
fitted_coef = model.coef_[0]
fitted_intercept = model.intercept_
print(f"y = {fitted_coef:.2f} * x + {fitted_intercept:.2f}")

y_pred = model.predict(X)

plt.figure(figsize=(8, 5))
plt.scatter(X, y, label="Data")
plt.plot(X, y_pred, color="red", label="Fitted line")
plt.xlabel("X")
plt.ylabel("y")
plt.title("x(1 dimension) and y")
plt.legend()
plt.show()
```

## Prediction on x

```python
#computing mse
def compute_mse(y_true, y_pred):
    mse = np.mean((y_true - y_pred) ** 2)
    return mse

mse_10 = compute_mse(y, y_pred)
print(f"MSE (noise=10): {mse_10:.4f}")
```

```python
# Repeat with the noise
X_50, y_50, coef_50 = make_regression(
    n_samples=100,
    n_features=1,
    n_informative=1,
    noise=50,
    coef=True,
    random_state=0,
    bias=100.0
)

model_50 = LinearRegression()
model_50.fit(X_50, y_50)

print(f"y = {model_50.coef_[0]:.2f} * x + {model_50.intercept_:.2f}")

y_pred_50 = model_50.predict(X_50)
mse_50 = compute_mse(y_50, y_pred_50)
print(f"MSE (noise=50): {mse_50:.4f}")
```
